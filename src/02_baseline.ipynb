# Majority-class baseline

This baseline always predicts the most common class in the training set. It gives us a simple reference point before fitting more useful models.

Run from the repository root or `src/`. We use the existing random split without call duration; preprocessing was fitted on the training rows. The positive class is a subscription (`y = 1`).

```python
from pathlib import Path
import sys

import pandas as pd
from sklearn.dummy import DummyClassifier

cwd = Path.cwd()
root = cwd.parent if cwd.name == "src" else cwd
sys.path.insert(0, str(root / "src"))
from model_utils import evaluate, load_random_split, validate_splits

splits = load_random_split(root / "data" / "processed" / "arrays.npz")
validate_splits(splits)
X_train, y_train = splits["train"]
baseline = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
print(f"Most common training class: {baseline.classes_[baseline.class_prior_.argmax()]}")
```

## Results

Average precision (PR-AUC) is our primary metric. For a constant score it equals the positive rate, while ROC-AUC is 0.5. Accuracy can look high simply because most customers did not subscribe. Precision, recall, and F1 use a 0.5 threshold.

```python
baseline_results = pd.DataFrame([
    {"model": "Majority", "split": name, **evaluate(baseline, X, y)}
    for name, (X, y) in splits.items()
])
baseline_results.round(4)
```
